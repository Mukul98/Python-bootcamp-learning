```python
## Declaring and Assigning Variables

a = 100
height = 5.9
name = "John Doe"
is_student = True

## Printing Variable Values
print("Value of a:", a)
print("Height:", height)
print("Name:", name)
print("Is Student:", is_student)
```

```python
## Naming Conventions
## Variable names should be descriptive and follow certain conventions. 
## For example, use lowercase letters and underscores for variable names (e.g., my_variable). Avoid using reserved keywords as variable names.
## must always start with a letter or underscore, and can contain letters, numbers, and underscores. Variable names are case-sensitive.

## valid variable names
my_variable = 10
first_name = "Alice"
last_name = "Smith"
```

```python
## Understanding Variable types
## Python is a dynamically typed language, which means that you don't have to explicitly declare the type of a variable. 
##The type is inferred based on the value assigned to it.

age = 25 #integer
name = "Alice" #string
height = 5.8 #float
is_student = True #boolean

print(type(age)) # Output: <class 'int'>
print(type(name)) # Output: <class 'str'>   
```

```python
## Type Checking and Type Conversion
# checking the type of variables
print(type(age)) # Output: <class 'int'>
```

```python
age = 25
print(type(age))  # Output: <class 'int'>

age_str = str(age)  # Convert integer to string
print(type(age_str))  # Output: <class 'str'>
```

```python
age = "25"
print(type(int(age)))  # Output: <class 'int'>
```

```python
## Dynamic Typing
# Python allows you to change the type of a variable by assigning a new value of a different type to it. This is known as dynamic typing.
x = 10  # x is an integer
print(type(x))  # Output: <class 'int'>
x = "Hello"  # x is now a string
print(type(x))  # Output: <class 'str'>
```

```python
## input() function
# The input() function is used to take user input in Python. 
# It reads a line from the input (usually from the user) and returns it as a string. You can also provide a prompt message to guide the user.

age = input("Enter your age: ")  # User inputs their age
print("You entered:", age)
```
