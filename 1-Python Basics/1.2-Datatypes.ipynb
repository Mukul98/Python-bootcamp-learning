```python
## Datatypes in Python
## Python has several built-in data types that are used to represent different kinds of data. Here are some of the most commonly used data types:
```

```python
## Intergers (int): Whole numbers without a decimal point. Example: 5, -3, 42
age = 25
type(age)  # Output: <class 'int'>
```

```python
## Floating-point numbers (float): Numbers with a decimal point. Example: 3.14, -0.001, 2.0
pi = 3.14
type(pi)  # Output: <class 'float'>
```

```python
## strings (str): A sequence of characters enclosed in single or double quotes. Example: "Hello", 'Python'
name = "Mukul"
type(name)  # Output: <class 'str'>
```

```python
## Booleans (bool): Represents one of two values: True or False. Example: True, False
is_student = True
type(is_student)  # Output: <class 'bool'>
```
