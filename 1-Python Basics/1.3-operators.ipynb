```python
## Operators in Python
## Arithemetic Operators
a=10
b=3

result_add = a + b  # Addition
print("Addition:", result_add)

result_sub = a - b  # Subtraction
print("Subtraction:", result_sub)

result_mul = a * b  # Multiplication
print("Multiplication:", result_mul)

result_div = a / b  # Division
print("Division:", result_div)

result_floor_div = a // b  # Floor Division
print("Floor Division:", result_floor_div)

modulus = a % b  # Modulus
print("Modulus:", modulus)

exponent = a ** b  # Exponentiation
print("Exponentiation:", exponent)
```

```python
## comparison Operators
x = 10
y = 20

print(x == y)  # Equal
print(x != y)  # Not Equal
print(x > y)   # Greater Than
print(x < y)   # Less Than
print(x >= y)  # Greater Than or Equal To
print(x <= y)  # Less Than or Equal To
```

```python
## Logical Operators
a = True
b = False

print(a and b)  # Logical AND
print(a or b)   # Logical OR
print(not a)    # Logical NOT
```
