```python
## Basic Syntax Rules in Python
## Case Sensitivity
name = "Mukul"
Name = "Malhotra"

print(name)
print(Name)
```

```python
## Indentation
## Indentation is very important in Python. It is used to define the blocks of code. For example, in an if statement, 
## the code block that follows the if condition must be indented.

age=18
if age >= 18:
    print("You are eligible to vote.")
```

```python
## Line Continuation
## In Python, you can use the backslash (\) character to indicate that a statement continues on the next line.

total = 1 + 2 + 3 + \
        4 + 5 + 6 + \
        7 + 8 + 9
    
print(total)
```

```python
## Multiple Statements on a Single Line
## You can write multiple statements on a single line using a semicolon (;) to separate them

x=10; y=20; z=30
print(x, y, z)
```

```python
## understanding Semantics in Python
# variable assignment
x = 5 # x is an integer
name = "John" # name is a string
```

```python
## Type Inference
# Python is a dynamically typed language, which means that you don't have to explicitly declare the type
print(type(x)) # Output: <class 'int'>
print(type(name)) # Output: <class 'str'>
```
